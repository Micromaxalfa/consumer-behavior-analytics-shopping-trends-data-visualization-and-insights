```python
import pandas as pd
import plotly.graph_objects as go

# Read the CSV from URL
url = "https://raw.githubusercontent.com/CSSEGISandData/COVID-19/master/archived_data/archived_daily_case_updates/02-09-2020_1030.csv"
df = pd.read_csv(url).head(10)
print(df.head())
# Display first 10 rows

# Get value counts for 'Last Update' column (shows date and time)
last_update_counts = df['Last Update'].value_counts()
labels = last_update_counts.index
counts = last_update_counts.values

# Create pie chart with Plotly
figure = go.Figure(data=[go.Pie(labels=labels, values=counts)])
figure.update_layout(title_text="Distribution of Rating by Last Update")
figure.update_traces(hoverinfo="label+percent", textinfo="value", textfont_size=14, marker=dict(line=dict(color='#000000', width=3)))

# Show figure
figure.show()
```

```python
# Making a BAr Chart using Plotly to konw the total purchase amount by Category
import pandas as pd
import plotly.express as px

# Load the CSV
url = "https://raw.githubusercontent.com/SaurabhSSB/consumer-behavior-analytics/main/shopping_trends.csv"
data = pd.read_csv(url).head(10)

# Bar chart of purchases by Category with custom color
fig = px.bar(
    data,
    x="Category",                # Categorical column
    y="Purchase Amount (USD)",   # Numeric column
    title="Total Purchase Amount by Category",
    color_discrete_sequence=["#F00808"]  # Change bar color (redish)
)

fig.show()
```

Data Analysis Project Summary
Data Source

The dataset utilized in this analysis, shopping_trends.csv, was sourced from a public GitHub repository. It provides a comprehensive snapshot of customer shopping behaviors, covering variables such as purchase amounts, product categories, customer demographics, locations, and rating preferences. This dataset serves as a valuable foundation for understanding consumer behavior patterns, evaluating purchasing trends, and generating actionable insights for business decision-making.

Data Exploration

To establish a clear understanding of the dataset, an initial exploratory data analysis (EDA) was conducted:

Data Structure & Integrity:

Verified null values and non-null entries across the dataset to ensure quality and completeness.

Checked data types and dimensions of the DataFrame to validate compatibility for analysis.

Descriptive Statistics:

Generated statistical summaries such as mean, median, mode, min, and max values for key variables.

Identified the maximum customer age to better understand the demographic distribution.

Reviewed the highest customer IDs, which provided a sense of dataset range and potential sample size.

DataFrame Overview:

Employed .info() and .describe() functions to review column types and data distribution.

Inspected categorical variables (e.g., color, size, category, location) to gauge diversity in customer behavior.

This exploration provided the groundwork for targeted visualizations and deeper analysis.

Visualizations

Multiple visualization techniques were applied to transform raw data into meaningful business insights:

Bar Charts:

Purchase Amount by Category: A customized bar chart illustrated which product categories generate the highest total sales. This enables strategic prioritization of product lines.

Purchase Amount by Location: A separate chart highlighted regional differences in purchase behavior, revealing high-performing geographic segments.

Customer Demographic Variables: Using both Matplotlib and Seaborn, additional bar charts captured counts and distributions for attributes such as customer ID, preferred color, clothing size, rating distribution, and seasonality of purchases.

Pie Chart:

Displayed the proportional distribution of product categories, offering a quick snapshot of which categories dominate in customer interest and sales share.

Line Plot:

Tracked the number of ratings across customer IDs, helping identify customer engagement levels and feedback trends. This visualization is particularly useful for monitoring satisfaction and loyalty.

Key Insights

From the analysis and visualizations, several insights emerged:

Category Performance: Certain product categories consistently outperformed others in terms of purchase volume and revenue contribution.

Regional Behavior: Purchase behavior varied significantly across locations, suggesting potential opportunities for targeted marketing campaigns or localized product offerings.

Customer Demographics: By analyzing color, size, and seasonal preferences, businesses can align inventory and marketing strategies with consumer demand.

Engagement Indicators: Customer ratings provided visibility into satisfaction levels, helping identify areas for product or service improvements.

Overall, these insights enable more data-driven decision-making for product strategy, inventory planning, and customer engagement.
